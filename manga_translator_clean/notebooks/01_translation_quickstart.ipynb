# Manga Translation Pipeline - Quick Start

This notebook demonstrates how to use the manga translation pipeline for translating manga pages.

## 1. Setup and Imports

```python
import sys
sys.path.insert(0, '..')

from src.pipeline import MangaTranslationPipeline
from config.settings import Settings
from PIL import Image
import matplotlib.pyplot as plt
```

## 2. Configure Pipeline

```python
# Create settings
settings = Settings()
settings.translator_type = 'google'  # Options: google, deepl, gemma, offline
settings.target_language = 'en'

# Initialize pipeline
pipeline = MangaTranslationPipeline(settings)
print("Pipeline initialized successfully!")
```

## 3. Translate a Single Image

```python
# Specify input and output paths
input_path = '../data/raw/sample_page.jpg'  # Update with your image path
output_path = '../data/processed/translated_page.jpg'

# Process image
result = pipeline.process_image(input_path, output_path)

print("Translation complete!")
print(f"Results: {result}")
```

## 4. Visualize Results

```python
# Load and display images
original = Image.open(input_path)
translated = Image.open(output_path)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 8))

ax1.imshow(original)
ax1.set_title('Original')
ax1.axis('off')

ax2.imshow(translated)
ax2.set_title('Translated')
ax2.axis('off')

plt.tight_layout()
plt.show()
```

## 5. Batch Processing

```python
from pathlib import Path

# Process multiple images
input_dir = Path('../data/raw')
output_dir = Path('../data/processed')
output_dir.mkdir(exist_ok=True)

image_files = list(input_dir.glob('*.jpg')) + list(input_dir.glob('*.png'))

for img_file in image_files:
    output_file = output_dir / f"translated_{img_file.name}"
    print(f"Processing: {img_file.name}...")
    
    try:
        pipeline.process_image(str(img_file), str(output_file))
        print(f"  ✓ Completed")
    except Exception as e:
        print(f"  ✗ Error: {e}")

print("\nBatch processing complete!")
```

## 6. Test Different Translators

```python
# Compare different translation engines
translators = ['google', 'offline']  # Add 'deepl', 'gemma' if configured

for translator in translators:
    print(f"\nTesting {translator} translator...")
    settings.translator_type = translator
    pipeline = MangaTranslationPipeline(settings)
    
    output_file = f'../data/processed/translated_{translator}.jpg'
    result = pipeline.process_image(input_path, output_file)
    print(f"  Completed: {result}")
```
